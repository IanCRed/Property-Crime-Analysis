Creating new column where high vs medium/low crime

```python
import numpy as np

df["Newcolumn"] = np.where(df["Property Crime Rate"] > 4000, "High Crime", "Medium or low Crime")
df.head()
```

Removing new column 

```python

df = df.drop(columns=["Newcolumn"])
```

Creating new column with 3 choices. 

```python
conditions = [
    (df["Property Crime Rate"] >= 2500), 
    (df["Property Crime Rate"] <2500) & (df["Property Crime Rate"] >= 1500),
    (df["Property Crime Rate"] < 1500),
]

choices = ["High Crime", "Medium Crime", "Low Crime"] 

df["Crime Level"] = np.select(conditions, choices, default="Unknown")

df.head()
```

Another faster way to add a new column with IFELSE 

```python
df["Gini Level"] = df["Gini"].apply(lambda x: "High Inequality" if x > 0.45 else "Medium Inequality" if x > 0.30 else "Low Inequality")
df.head()
```

Creating a new column using if elif else

```python
def UnemploymentLevel(row):
    if row["Unemp_Rate"] > 7:
        return "High Unemployment"
    elif row["Unemp_Rate"] > 4:
        return "Medium Unemployment"
    else:
        return "Low Unemployment"
    
df["Unemployment Level"] = df.apply(UnemploymentLevel, axis=1)

df.head()
```
